```python
#Manual uploading of the file from the computer
from google.colab import files
uploaded=files.upload()
rtf_file_name=list(uploaded.keys())[0]
print("Uploaded file:", rtf_file_name)
```

Convert RTF to Plain Text (to reduce the no of tokens passed to the LLM)

```python
#this function will convert it to text, but the table structure will disappear
from striprtf.striprtf import rtf_to_text

with open(rtf_file_name, "r", encoding="utf-8", errors="ignore") as f:
  rtf_content=f.read()
plain_text=rtf_to_text(rtf_content)
print(plain_text[:2000])
```

```python
import re
lines=plain_text.splitlines()
clean_lines=[]

for line in lines:
  line=line.strip()
  if not line: #skip empty line
    continue
  # Skip page numbers
  if re.search(r"Page\s+\d+\s+of\s+\d+", line):
      continue

  # Skip protocol header
  if "ZYLiQ Protocol" in line:
      continue

  # Skip source footer
  if line.startswith("Source:"):
      continue

  # Skip program footer
  if line.startswith("Program:"):
      continue

  clean_lines.append(line) #keep everything else
clean_text="\n".join(clean_lines) #join() takes an iterable of strings (usually a list of strings) and combines them into a single string.
print(clean_text[:2000])
```

Load the Extractor Prompt file which is in GitHub

```python

with open("prompts/extractor_prompt.txt", "r", encoding="utf-8") as f:
  prompt_template=f.read()
final_prompt=prompt_template.replace("{document}",clean_text)
print(final_prompt[:3000])
```
